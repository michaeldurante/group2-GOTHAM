```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("tess_data.csv")
n_bins = 1000

flux_cols = [f'flux_{i:04d}' for i in range(n_bins)]
flux_err_cols = [f'flux_err_{i:04d}' for i in range(n_bins)]
X = df[flux_cols].values
X_err = df[flux_err_cols].values

y = df['label'].values
X = X[y == 1]
X_err = X_err[y == 1]
# plt.errorbar(range(n_bins), X[1], yerr=X_err[1], fmt='o', markersize=2, capsize=2)
# plt.show()  
# plot_data = data.iloc[:, 1:].values
# plt.plot(plot_data)
# plt.show()
```

```python
idx = 1
print(np.shape(X))
plt.figure(figsize = (15,3))
# plt.errorbar(range(n_bins), X[idx], yerr=X_err[idx], fmt='o', markersize=2, capsize=2)
# plt.scatter(range(n_bins), X[idx])
plt.plot(range(n_bins), X[idx])
plt.fill_between(range(n_bins),
                X[idx] - X_err[idx],
                X[idx] + X_err[idx],
                color="tab:blue",
                alpha=0.22,
                label="±1σ uncertainty",
            )
# plt.xlim(600,700)
plt.show()  
```
